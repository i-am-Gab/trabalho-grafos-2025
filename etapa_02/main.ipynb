1. Importação dos arquivos de código:
Com os trechos de códgio estão organizados em arquivos e pastas dentro do projeto, fazemos a importação de todos os arquivos necessários.

```python
import sys
import os

sys.path.append(os.path.abspath('etapa_02'))

diretorio = 'app/storage/data/arquivos'
caminho_saida = 'app/storage/solucoes'

CLOCKS_PER_SEC = 1_000_000

from app.src.utils.utils import escolher_arquivos
from app.src.utils.single_data import executar_algoritmo_arquivo_unico
from app.src.utils.multiple_data import executar_algoritmo_arquivos_multiplos
```

```python
resultado = escolher_arquivos(diretorio)
```

```python
if resultado is None:
    print("Nenhum arquivo selecionado. Encerrando.")
    
elif isinstance(resultado, list):
   executar_algoritmo_arquivos_multiplos(resultado, caminho_saida, CLOCKS_PER_SEC)
        
else:
    executar_algoritmo_arquivo_unico(resultado, caminho_saida, CLOCKS_PER_SEC)
```
