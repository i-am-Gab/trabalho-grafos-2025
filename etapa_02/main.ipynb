1. Importação dos arquivos de código:
Com os trechos de códgio estão organizados em arquivos e pastas dentro do projeto, fazemos a importação de todos os arquivos necessários.

```python
import sys
import os

sys.path.append(os.path.abspath('etapa_02'))

diretorio = 'app/storage/data/arquivos'

from app.src.loader.loader import carregar_instancia
from app.src.utils.utils import escolher_arquivos
```

```python
if resultado is None:
    print("Nenhum arquivo selecionado. Encerrando.")
    
elif isinstance(resultado, list):
    for caminho in resultado:
        grafo = carregar_instancia(caminho)
else:
    caminho = resultado
    grafo = carregar_instancia(caminho)
```
